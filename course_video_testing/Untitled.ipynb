# Filling the Gaps: Helping Boilerexams Content Creators Identify Missing Video Explanations  

## Introduction and Problem  

On the **Boilerexams** website, several exams lack video explanations. While the absence of these videos may seem insignificant, they play a crucial role in enhancing the study experience for students using Boilerexams. Video explanations serve as valuable tools to clarify doubts and alleviate pre-exam anxiety, especially when assistance from course staff, such as TAs or professors, is not readily available.

The gap of explainer videos became apparent to me when I was preparing for CS251 exams. Late at night, I often encountered difficulties understanding certain concepts or solving practice problems on the site. I would ask my classmates, but they frequently had the same questions or offered conflicting explanations about why certain answer choices were incorrect. During these hours, responses from TAs on platforms like Ed were limited, leaving many questions unanswered. In contrast, while studying for a Linear Algebra exam, I noticed that the availability of explainer videos associated with practice questions helped me correct misunderstandings more effectively. Experiencing these differences firsthand motivated me to develop a tool that identifies which questions are missing videos, allowing the content team to address the gap of missing videos, in order to improve the platform.

---

### Methodology
The content team is made up of Purdue students; as a result, they only have so much time. Therefore, if we wanted to both help students and make good use of the content team's time, we needed to **identify the most optimal questions**.  

We came up with three metrics to prioritize which questions needed videos:
1. Average percent correct – How often students got the question right.
2. Total submissions – How many times the question was attempted.
3. Number of correct submissions – How many of those attempts were correct.

### Building the Table  
To build the table, we needed to gather data from multiple tables in the database. We combined the Questions, Exam, Course, and Video tables to build a complete list of questions without videos.

After exploring the large table with question, exam, course, and video data, we discovered that some questions were associated with multiple topics. For example, a shell sorting question might be under both Sorting and Shell Sorting. To clean this up, we grouped all the duplicate questions together so that each question appeared only once, but with all its associated topics listed.

![Table](table.png)

With the table ready, we moved on to making the tool useful for the content team.

### Building the Interface
A static table was inefficient for the content team to use, as it took a lot of searching to find questions for a single course. Therefore, I built an interactive interface using Jupyter Notebook’s ipywidgets. The interface lets the team:  
- Select a course they’re interested in.  
- Sort questions by any of the three metrics to focus on the most important questions.
- Click a link to a question to get context on said question

## What’s Next?
This tool is a start. However, there are some further steps that can be taken to improve the product. Here are a few ideas:
1. Integration with the content dashboard – Embedding the tool directly into the course creation page so the content team doesn’t have to load up the notebook.
2. Real-Time Insights on Most-Missed Questions – Adding a feature that tracks a running average of the most-missed questions leading up to an exam. This would be helpful for:
   - courses with fewer exams or that are newer, the tool helps prioritize missing videos for upcoming exams based on recent student struggles.
   - courses with many exams, it focuses on aligning video creation with current student needs by analyzing recent data.

---

## Why This Matters?
This tool is something that can **augment the content team's time and effort** by giving them the data they need to make smart decisions about where to focus their work. More importantly, this tool aims to reduce the number of questions missing video explanations — something I wish I had when I was taking CS251.
